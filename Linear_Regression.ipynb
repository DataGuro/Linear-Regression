# Simple Linear Regression

```python
import pandas as pd

df = pd.read_csv("delivery_time.csv")
df.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df.distance_km, df.delivery_time_min, color='#5BB8E8', edgecolor='k')
plt.xlabel("Distance (km)")
plt.ylabel ("Delivary Time (min)")
plt.title("Father orders take longer... looks like a line!")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(df[['distance_km']], df['delivery_time_min'])
```

```python
model.predict(pd.DataFrame({'distance_km': [7, 34]}))
```

```python
model.coef_
```

```python
model.intercept_
```

```python
'''
y = m*x + b
time = m * distance + b
'''

time = 2.81819063 * 7 + 10.944
time 
```

```python
plt.scatter(df.distance_km, df.delivery_time_min, color='#5BB8E8', edgecolor='k')
plt.plot(df.distance_km, model.predict(df[['distance_km']]), color='#39D353', linewidth=2)
plt.xlabel("Distance (km)")
plt.ylabel("Delivery Time (min)")
plt.title("Best fit line: time = m * distance + b")
plt.show()
```

# Multiple Linear Regression

```python
df2 = pd.read_csv("delivery_time_multi.csv")
df2.head()
```

```python
model2 = LinearRegression()

features = ['distance_km', 'traffic_level', 'raining']

model2.fit(df2[features], df2['delivery_time_min']) #training
```

```python
new_order = pd.DataFrame({'distance_km': [7], 'traffic_level': [8], 'raining': [1]})
predicted = model2.predict(new_order) # inference;
```
